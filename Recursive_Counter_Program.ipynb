```
import sys

# Ensure input function compatibility across common Python versions
# In modern Python 3, 'input()' is used.
# Since Python 2 is largely deprecated, we assume Python 3's 'input'.

def countdown(n):
    """
    Counts down from a positive integer n until 0, then prints 'Blastoff!'.
    """
    if n <= 0:
        print('Blastoff!')
    else:
        # Use int(n) to ensure clean output if the input was float-converted earlier
        print(int(n))
        countdown(n - 1)

def countup(n):
    """
    Counts up from a negative integer n until 0, then prints 'Blastoff!'.
    """
    # The base case is reached when n is 0 or greater
    if n >= 0:
        print('Blastoff!')
    else:
        # Use int(n) for clean output
        print(int(n))
        countup(n + 1) # Move towards the base case (0) by incrementing

def main():
    """
    Handles user input and directs the flow to either countdown or countup.
    """
    print("--- Recursive Counter Program ---")
    try:
        # Get input as a string and attempt conversion to an integer
        user_input_str = input("Enter an integer: ")
        num = int(user_input_str)

        if num > 0:
            print("\n--- Output for Positive Input ---")
            print(f"Calling countdown({num}):")
            countdown(num)
        elif num < 0:
            print("\n--- Output for Negative Input ---")
            print(f"Calling countup({num}):")
            countup(num)
        else: # num == 0
            print("\n--- Output for Zero Input ---")
            print(f"Calling countdown({num}):")
            countdown(num)

    except ValueError:
        print("\nError: Invalid input. Please ensure you enter a valid integer.")

if __name__ == "__main__":
    main()
```
